# Converter

This notebook is to convert the data in `data/` into valid insert statements in the oracle dialect

```python
import re
import json

import pandas as pd
```

```python
data = pd.read_csv("data/pincodes.csv", dtype={'district': str, 'pincode': int})
```

```python
data = data[[ "officename", "pincode", "district", "statename"]]
data["statename"] = data["statename"].astype(str).apply(str.title)
data.drop(data[data["statename"] == "Nan"].index, inplace=True)


data["district"] = data["district"].astype(str).apply(str.title)
data.drop(data[data["district"] == "Nan"].index, inplace=True)
```

```python
states = data.drop_duplicates("statename")[["statename"]]
states.to_csv('data/states.csv', index=False)
```

```python
districts = data.drop_duplicates("district")[["district", "statename"]]
districts.to_csv('data/districts.csv', index=False)
```

```python

pincodes = data.drop_duplicates("pincode")[["pincode", "district"]]
pincodes.to_csv('data/addresses.csv', index=False)
```

```python
offices = data.copy()[["officename", "pincode"]]

suffix = re.compile(r'([PBHS].??[Oo])$')

offices["officename"] = offices["officename"].astype(str).apply(lambda name: re.sub(suffix, '', name.strip()).replace(".", " ").strip() +" PDC")

offices.to_csv('data/distributors.csv', index=False)
```
